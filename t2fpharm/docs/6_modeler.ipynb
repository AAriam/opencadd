# Pharmacophore Modeler

The `t2fpharm.modeler.Modeler` object provides three methods
to perceive pharmacophores from the input field and pocket data: `Modeler.simple()`, `Modeler.cnn()`, and `Modeler.largest_peaks`.

## Simple Method

This is the core method that is also used by the `cnn` and `largest_peaks` methods.
The algorithm works as follows:

For each feature type in the field,
1. If `filter_function` is provided,
   for each grid point the field value is replaced
   by the value of the `filter_function`
   applied to the field values within a sphere
   of radius `filter_radius` centered at the grid point.
   This can be used to smoothen the field
   (e.g., using a Gaussian or mean filter),
   or to apply a custom transformation to the field values.
2. If `best_per_point` is `True`,
   grid points whose field values (after applying the filter function, if any)
   are not the best (i.e. lowest for "min" peaks or highest for "max" peaks)
   among all feature type field values at that grid point are discarded.
   This prevents selecting the same grid point for multiple feature types.
3. If `threshold_value` is provided,
   grid points whose field values
   (after applying the filter function, if any)
   do not meet the threshold are discarded.
   This is useful to filter out noise
   and reduce the number of features.
4. If the modeler has a pocket defined,
   grid points outside the pocket are discarded.
   This ensures that only features
   within the pocket are considered.
5. If `threshold_percentile` is provided,
   only the best `threshold_percentile` percent
   of the remaining grid points
   are kept for each feature type.
6. The remaining grid points are taken as pharmacophore feature centers
   and used to create a `Pharmacophore` object.

More information is available in the method's docstring:

```python
import t2fpharm
help(t2fpharm.Modeler.simple)
```

## CNN Method

The `cnn` method starts in the same way as the `simple` method,
only afterwards the generated features are clustered together using the CNN algorithm
with the provided parameters.
Thus, calling `Modeler.cnn()` is equivalent to calling `Modeler.simple()`
to create an initial pharmacophore,
followed by calling `Pharmacophore.cluster_cnn()` on the generated `Pharmacophore` object
to cluster the feature centers.
For more information on the algorithm and parameters,
see the documentation of those methods.

```python
help(t2fpharm.pharm.Pharmacophore.cluster_cnn)
```

```python
help(t2fpharm.pharm.Pharmacophore.cluster)
```

## Largest Peaks Method

The `largest_peaks` method starts in the same way as the `simple` method,
only afterwards the generated features are filtered based on their distances and priorities.
Thus, calling `Modeler.largest_peaks()` is equivalent to calling `Modeler.simple()`
to create an initial pharmacophore,
followed by calling `Pharmacophore.remove_overlaps()` on the generated `Pharmacophore` object
to filter the feature centers.
For more information on the algorithm and parameters,
see the documentation of those methods.

```python
help(t2fpharm.pharm.Pharmacophore.remove_overlaps)
```
