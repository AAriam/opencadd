# Quickstart Guide

All functionalities are accessible from the top-level `t2fpharm` package,
which is the only module you need to import:

```python
import t2fpharm
```

A T<sup>2</sup>F pharmacophore modeler can then be created using the `t2fpharm.modeler()` function:

```python
help(t2fpharm.modeler)
```

To create a new modeler,
at the very least you need a set of 3D scalar fields,
each measuring some metric for the favorability of a pharamacophore feature
at different points within the binding pocket of a receptor.
The modeler then provides various methods
to extract pharmacophore features from the corresponding field values.
If the fields do not already contain information about the binding pocket,
you may also optionally provide pocket information so that only those
field datapoints that lie within the pocket are considered for pharmacophore generation.
Moreover, you may optionally provide the receptor itself as well,
which can then be used in post-analysis steps, e.g., for visualization.
Each of these inputs must be provided as a special object,
which can be created from various inputs as explained below.

## Grid Generation

To create a field from any of the available methods,
you first need to specify the spatial grid
on which the field is sampled.
If you opt in to provide the modeler with binding pocket information as well
(e.g., because your field data points exceed the pocket boundaries and thus need to be masked),
then you can (and must) simply reuse the same `Grid` object associated with the pocket to generate a field (see section Pocket Generation).
Otherwise, you must manually define your grid using one of the available functions in the `t2fpharm.grid` module, as described below.

### From Anchor Point, Shape, and Size

You can define a `Grid` by its shape, size,
and the coordinates of an anchor point,
using the `t2fpharm.grid.from_anchor_shape_size()` function:



```python
help(t2fpharm.grid.from_anchor_shape_size)
```

For example:

```python
t2fpharm.grid.from_anchor_shape_size(
    anchor=(0, 0, 0),
    shape=(10, 20, 30),
    size=(1, 2, 3),
)
```

### From Anchor Point, Shape, and Spacing

You can define a `Grid` by its shape, spacing,
and the coordinates of an anchor point,
using the `t2fpharm.grid.from_anchor_shape_spacing()` function:

```python
help(t2fpharm.grid.from_anchor_shape_spacing)
```

For example:

```python
t2fpharm.grid.from_anchor_shape_spacing(
    anchor=(0, 0, 0),
    shape=(10, 20, 30),
    spacing=0.5,
)
```

### From Anchor Point, Size, and Spacing

You can define a `Grid` by its size, spacing,
and the coordinates of an anchor point,
using the `t2fpharm.grid.from_anchor_size_spacing()` function:

```python
help(t2fpharm.grid.from_anchor_size_spacing)
```

For example:

```python
t2fpharm.grid.from_anchor_size_spacing(
    anchor=(0, 0, 0),
    size=(10, 20, 30),
    spacing=(0.5, 1, 1.5),
    anchor_type="center",
)
```

### From Bounds and Shape

You can define a `Grid` by its bounds and shape,
using the `t2fpharm.grid.from_bounds_shape()` function:

```python
help(t2fpharm.grid.from_bounds_shape)
```

For example:

```python
t2fpharm.grid.from_bounds_shape(
    lower=(-5, -10, -15),
    upper=(3.5, 7.5, 10),
    shape=(10, 20, 30),
)
```

### From Bounds and Spacing

You can define a `Grid` by its bounds and spacing,
using the `t2fpharm.grid.from_bounds_spacing()` function:

```python
help(t2fpharm.grid.from_bounds_spacing)
```

For example:

```python
t2fpharm.grid.from_bounds_spacing(
    lower=(0, 0, 0),
    upper=(10, 10, 10),
    spacing=0.3,
)
```

## Field Generation

### From Pre-Computed Data

The shortest and most flexible way to create a `Field` object
is by providing your own pre-computed field data to the `t2fpharm.field.from_data()` function:

```python
help(t2fpharm.field.from_data)
```

For example:

```python
import numpy as np

# Dummy field data for 5 feature metrics,
# each sampled on a 20x30x40 grid:
feature_ids = ["f1", "f2", "f3", "f4", "f5"]
field_data = np.random.rand(len(feature_ids), 20, 30, 40)

# Suppose the field's grid center is at (15.5, 16, 17.5)
# and grid spacing is 0.3 in all three dimensions:
field_grid = t2fpharm.grid.from_anchor_shape_spacing(
    anchor=(15.5, 16, 17.5),
    shape=field_data.shape[1:],
    spacing=0.3,
    anchor_type="center",
)

t2fpharm.field.from_data(
    grid=field_grid,
    tensor=field_data,
    feature_ids=feature_ids,
)
```
