# Grid Generation

To create a field from any of the available methods,
you first need to specify the spatial grid
on which the field is sampled.
If you opt in to provide the modeler with binding pocket information as well
(e.g., because your field data points exceed the pocket boundaries and thus need to be masked),
then you can (and must) simply reuse the same `Grid` object 
associated with the pocket to generate a field.
Otherwise, you must manually define your grid
using one of the available functions in the `t2fpharm.grid` module, as described below.

```python
import t2fpharm
```

## From Anchor Point, Shape, and Size

You can define a `Grid` by its shape, size,
and the coordinates of an anchor point,
using the `t2fpharm.grid.from_anchor_shape_size()` function:



```python
t2fpharm.grid.from_anchor_shape_size(
    anchor=(0, 0, 0),    # Coordinates of the anchor point (in the same length unit as other data, usually Å)
    shape=(10, 20, 30),  # Shape of the grid
    size=(1, 2, 3),      # Physical size of the grid (in the same length unit as `anchor`)
    anchor_type="lower"  # Type of the anchor point; either "lower", "center", or "upper"
)
```

## From Anchor Point, Shape, and Spacing

You can define a `Grid` by its shape, spacing,
and the coordinates of an anchor point,
using the `t2fpharm.grid.from_anchor_shape_spacing()` function:

```python
t2fpharm.grid.from_anchor_shape_spacing(
    anchor=(0, 0, 0),      
    shape=(10, 20, 30),    
    spacing=0.5,  # Spacing between grid points along main axes (in the same length unit as `anchor`)
    anchor_type="center",
)
```

## From Anchor Point, Size, and Spacing

You can define a `Grid` by its size, spacing,
and the coordinates of an anchor point,
using the `t2fpharm.grid.from_anchor_size_spacing()` function:

```python
t2fpharm.grid.from_anchor_size_spacing(
    anchor=(0, 0, 0),
    size=(10, 20, 30),
    spacing=0.5,
    anchor_type="center",
)
```

## From Bounds and Shape

You can define a `Grid` by its bounds and shape,
using the `t2fpharm.grid.from_bounds_shape()` function:

```python
t2fpharm.grid.from_bounds_shape(
    lower=(-5, -10, -15),  # Coordinates of the lower bounds (i.e. origin)
    upper=(3.5, 7.5, 10),  # Coordinates of the upper bounds
    shape=(10, 20, 30),
)
```

## From Bounds and Spacing

You can define a `Grid` by its bounds and spacing,
using the `t2fpharm.grid.from_bounds_spacing()` function:

```python
grid = t2fpharm.grid.from_bounds_spacing(
    lower=(0, 0, 0),
    upper=(10, 10, 10),
    spacing=0.3,
)

grid
```

For a set of given coordinates, find the index of the nearest grid point to each given coordinates,
their distance, and whether the coordinates lie inside the grid:

```python
indices, distances, is_inside = grid.nearest_point([[1, 0.5, 0.5], [20, 20, 20]])
print(f"indices: {indices}", f"distances: {distances}", f"is inside: {is_inside}", sep="\n")
```

Get coordinates of grid points given their indices:

```python
grid.index_coordinates([[0,0,0], [3,4,5]])
```

```python
grid.common_neighbors_count(neighborhood_radius=1.5, point1=(0,0,0), point2=(0,0,1))
```

Convert the grid to a dictionary, which can be saved and used to recreate the grid:

```python
grid_data = grid.to_dict()
grid_data
```

```python
grid_recreated = t2fpharm.grid.from_data(**grid_data)
```

Check for equality between two grids:

```python
grid == grid_recreated
```
